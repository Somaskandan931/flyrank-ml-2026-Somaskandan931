## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

**Task type: Classification.**

The task is to classify each content item as declining or not declining. The output supports identifying content that may need review or a refresh. Classification fits because the output is a discrete decision between two classes rather than a continuous value or an ordered list.

```python
!git clone https://github.com/Somaskandan931/flyrank-ml-2026-Somaskandan931.git
%cd flyrank-ml-2026-Somaskandan931
```

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# The label is defined from trend_direction in the reference pipeline.
df["is_declining_label"] = (df["trend_direction"] == "down").astype(int)

print("Dataset shape:", df.shape)
print("Target values:")
print(df["is_declining_label"].value_counts())
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target is `is_declining_label`. It is a defined label created from `trend_direction`, where a downward trend is labelled as declining. I will use this as the classification target. `trend_direction` and `trend_pct` must not be used as features because they are directly involved in defining the target and would create target leakage.

```python
print("Target column: is_declining_label")
print("\nTarget distribution:")
print(df["is_declining_label"].value_counts())

print("\nDeclining rate:")
print(df["is_declining_label"].mean())
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

**Primary metric: F1 score.**

F1 is appropriate because the goal is to identify declining content while balancing precision and recall. Accuracy alone could be misleading if the declining and non-declining classes are imbalanced. A higher F1 score means a better balance between finding declining content and avoiding false alerts.

```python
class_counts = df["is_declining_label"].value_counts()
class_proportions = df["is_declining_label"].value_counts(normalize=True)

print("Class counts:")
print(class_counts)

print("\nClass proportions:")
print(class_proportions)
```

## 4. The unit of analysis, as a real dataframe

**One row represents one pseudonymized content item.**

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could flag content using one manually chosen threshold, such as a drop in traffic or engagement. However, content performance depends on multiple signals and these patterns can differ across content and clients. ML can learn relationships across several signals instead of relying on one manually chosen threshold. The resulting classification can support the decision of which content should be reviewed or considered for a refresh.

```python
candidate_columns = [
    "ctr",
    "engagement_rate",
    "scroll_rate",
    "ai_traffic_pct",
    "avg_position",
    "word_count"
]

available_columns = [
    col for col in candidate_columns
    if col in df.columns
]

print("Available candidate signals:")
print(available_columns)

display(df[available_columns].head())
```
